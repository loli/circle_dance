# Audio Devices

List and select the devices to listen to.

```python
import pulsectl  # pulse audio lib
import pyaudio  # stream listening lib
import sounddevice as sd  # alterantive to pyaudio, also using ALSA
```

```python
with pulsectl.Pulse('my-client') as pulse:
    sources = pulse.source_list()
    for i, source in enumerate(sources):
        print(f"Device {i}:")
        print(source)
        #print(f"Source {i}: {source.name} - {source.description}")
```

```python
p = pyaudio.PyAudio()

# List all available devices
for i in range(p.get_device_count()):
    info = p.get_device_info_by_index(i)
    if info['maxOutputChannels'] > 0:
        print(f"Device {i}:")
        print(info)
        #print(f"  Name: {info['name']}")
        #print(f"  Max Input Channels: {info['maxInputChannels']}")
        #print(f"  Max Output Channels: {info['maxOutputChannels']}")
        #print(f"  Default Sample Rate: {info['defaultSampleRate']}")
        print()

p.terminate()
```

```python
p = pyaudio.PyAudio()
for i in range(p.get_device_count()):
    dev = p.get_device_info_by_index(i)
    if dev['name'].lower().find('pulse') >= 0:
        print(i)
        print(f"Found PulseAudio device: {dev['name']}")
        # Use this device index when opening the stream
        break
```
